Brief:  
There are many efficient methods for calculating the DCT of a discrete-time signal. A lot of these methods rely on the FFT algorithm. We're going to explore the various methods for a test signal x[n] = (1, 1, 1, 2).

Reference:
- https://dsp.stackexchange.com/a/10606

## Imports and Setup

```python
import numpy as np
from   scipy.fft import fft, dct
```

## Input Signal

```python
x_n = np.array([1,1,1,2], dtype=float)
N   = x_n.size
```

## Method 1: 2N-point FFT (DCT-II)

```python
# 2N-point even extension of x[n].
y_n = np.zeros(2*N)
y_n[0:N] = x_n
y_n[N:]  = x_n[::-1]

phase_shift = np.zeros(N, dtype=complex)
for k in range(N):
    phase_shift[k] = np.exp(-1j*np.pi*k/(2*N))

my_fft_1 = fft(y_n)
my_dct_1 = (phase_shift * my_fft_1[0:N]).real
print(f"result: {my_dct_1}")
```

## Method 2: Fast Cosine Transform

```python
# N-point signal derived from the even indices of the 2N-point even extension of x[n].
v_n = np.zeros(N)
for n in range(N):
    v_n[n] = y_n[2*n]

phase_shift = np.zeros(N, dtype=complex)
for k in range(N):
    phase_shift[k] = 2*np.exp(-1j*np.pi*k/(2*N))

my_fft_2 = fft(v_n)
my_dct_2 = (phase_shift * my_fft_2).real
print(f"result: {my_dct_2}")
```

## Method 3: 4N-point FFT with no phase shifts

```python
# 4N-point extension of x[n] with zeros at even indexes.
z_n = np.zeros(4*N)
i   = 0
for n in range(4*N):
    if(n%2 != 0):
        z_n[n] = y_n[i]
        i = i + 1

my_fft_3 = fft(z_n)
my_dct_3 = my_fft_3[0:round(z_n.size/4)].real
print(f"result: {my_dct_3}")
```

## Method 4: Using the built-in DCT function

```python
my_dct_4 = dct(x_n)
print(f"result: {my_dct_4}")
```
